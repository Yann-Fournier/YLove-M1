```python
from sentence_transformers import SentenceTransformer

# ===============================
# Données
# ===============================
textes = [
    "The weather is lovely today.",
    "It's so sunny outside!",
    "He drove to the stadium.",
]

# ===============================
# Modèle 1 : all-MiniLM-L6-v2
# ===============================
model_minilm = SentenceTransformer("all-MiniLM-L6-v2")

embeddings_minilm = model_minilm.encode(textes)

print("all-MiniLM-L6-v2")
print("Shape :", embeddings_minilm.shape)
print(embeddings_minilm)

# ===============================
# Modèle 2 : distiluse-base-multilingual-cased-v2
# ===============================
model_distiluse = SentenceTransformer("distiluse-base-multilingual-cased-v2")

embeddings_distiluse = model_distiluse.encode(textes)

print("\ndistiluse-base-multilingual-cased-v2")
print("Shape :", embeddings_distiluse.shape)
print(embeddings_distiluse)
```

```python
from sentence_transformers import SentenceTransformer

# Chargement du modèle SBERT
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

# Textes à vectoriser
textes = ["Le chat dort sur le tapis.", "Le chien joue dans le jardin."]

# Obtention des embeddings
embeddings_minilm = model.encode(textes)
print(embeddings_minilm.shape)
print(embeddings_minilm)
```

```python
import numpy as np

# Vecteurs des deux textes
vec1 = embeddings_minilm[0]
vec2 = embeddings_minilm[1]

print(vec1)
print(vec2)

# Calcul de la similarité cosinus
similarité = np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

print(f"Similarité : {similarité}")
```

```python
import pandas as pd

hobbys_csv_path = '../data/ChatGPT/liste_hobbies.csv'

hobbies_df = pd.read_csv(hobbys_csv_path)
hobbies = hobbies_df['Hobby'].tolist()
```

```python
# Vectorisation avec Sentence Transformers (recommandé)
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
model2 = SentenceTransformer("distiluse-base-multilingual-cased-v2") # Modèle multilingue (comme on est que en français ici)
vectors = model.encode(hobbies)
vectors2 = model2.encode(hobbies)

print(vectors.shape)
print(vectors2.shape)
# (9, 384)
```

```python
# Mesurer la similarité entre hobbies
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(vectors)
similarity_matrix2 = cosine_similarity(vectors2)

print(similarity_matrix)
print(similarity_matrix2)
```

```python
# from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# similarities = cosine_similarity(vectors)

def recommander(hobby, top_k=3):
    idx = hobbies.index(hobby)
    scores = similarity_matrix[idx]

    similar_indices = np.argsort(scores)[::-1][1:top_k+1]
    return [(hobbies[i], scores[i]) for i in similar_indices]
    

print(recommander("Jeux vidéo"))
```

```python
def recommander_2(hobby, top_k=3):
    idx = hobbies.index(hobby)
    scores2 = similarity_matrix2[idx]

    similar_indices_2 = np.argsort(scores2)[::-1][1:top_k+1]
    return [(hobbies[j], scores2[j]) for j in similar_indices_2]
    

print(recommander_2("Jeux vidéo"))
```

```python
# Projection en 2D pour visualisation
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
vectors_2d = pca.fit_transform(vectors)

pca_2 = PCA(n_components=2)
vectors_2d_2 = pca_2.fit_transform(vectors2)

```

```python
# Visualisation dans un plan
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(vectors_2d[:, 0], vectors_2d[:, 1])

for i, hobby in enumerate(hobbies):
    plt.annotate(hobby, (vectors_2d[i, 0], vectors_2d[i, 1]))
    


plt.title("Vectorisation des caractere - all-MiniLM-L6-v2")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(vectors_2d_2[:, 0], vectors_2d_2[:, 1])

for i, hobby in enumerate(hobbies):
    plt.annotate(hobby, (vectors_2d_2[i, 0], vectors_2d_2[i, 1]))

plt.title("Vectorisation des caractères – multi-qa-MiniLM-L6-cos-v1")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True)
plt.show()
```
